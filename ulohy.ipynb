# Úloha 1: Lineární a nelineární oscilátory

- Tento soubor se zaměřuje na fyzikální systémy, které kmitají.
- Porovnáváte zde dva typy modelů:

    1. **Lineární tlumený oscilátor**
         - Klasický model pružiny se závažím a tlumením.
         - Systém je předvídatelný.
         - Po určité době se díky tlumení ustálí v rovnovážné poloze.

    2. **Duffingův oscilátor (nelineární)**
         - Model přidává nelineární člen (třetí mocnina výchylky).
         - Zahrnuje vnější budicí sílu.
         - Simuluje deterministický chaos.
         - I malá změna počátečních podmínek vede k úplně jinému výsledku.

- **Co dělá**
    - Simuluje dynamiku nelineárního oscilátoru a chaos.

- **Využití**
    - Analýza fázových portrétů (vztah mezi výchylkou a rychlostí).
    - Pozorování podivných atraktorů.

- **Implementace**
    - Používáte knihovnu `scipy.integrate.odeint` pro numerické řešení rovnic druhého řádu.
    - Tyto rovnice jsou převedeny na soustavu dvou rovnic prvního řádu.

# Úloha 2: SIR model (epidemiologie)

- Tento skript simuluje šíření infekčních nemocí v populaci.
- Populace je rozdělena do tří skupin:

    - `S` (Susceptible): zdraví jedinci, kteří mohou onemocnět.
    - `I` (Infected): nakažení jedinci, kteří nemoc dále šíří.
    - `R` (Recovered/Removed): vyléčení nebo imunní jedinci, kteří již nemoc nešíří.

- **Klíčové parametry**
    - Reprodukční číslo ($R_0$): určuje, kolik lidí průměrně nakazí jeden nemocný.
    - Čím vyšší $R_0$, tím dříve nastává vrchol epidemie a tím více lidí celkově onemocní.
    - Doba trvání: u nemocí s vysokou nakažlivostí projde vlna populací rychleji.

# Úloha 3: Vlastní model (Zombie apokalypsa – SZR)

- Jedná se o modifikaci SIR modelu aplikovanou na fiktivní scénář „zombie nákazy“.
- Struktura je podobná, ale pravidla interakce jsou odlišná:

    - `S` (Susceptible): zdraví lidé.
    - `Z` (Zombies): nakažení jedinci, kteří aktivně „loví“ zdravé lidi.
    - `R` (Removed): „zneškodnění“ zombie.

- **Specifikum modelu**
    - Na rozdíl od klasického SIR modelu, kde se lidé v kategorii `R` prostě vyléčí, zde existuje možnost reanimace.
    - Mrtví zombie se mohou oživit zpět do kategorie `Z`, což činí model dynamičtějším a nebezpečnějším pro přežití lidstva.

- **Matematika**
    - Model je opět implementován jako soustava tří nelineárních diferenciálních rovnic řešených pomocí `odeint`.

# Technické minimum pro diskuzi

- Použité knihovny:
    - `numpy` pro pole a výpočty
    - `matplotlib.pyplot` pro vykreslování grafů
    - `scipy.integrate.odeint` pro řešení ODE

- Princip ODE řešiče:
    - Funkce `odeint` bere jako vstup:
        - funkci definující derivace,
        - počáteční stav systému,
        - časový vektor.
    - Výstupem je matice stavů v čase.

- Fázový prostor:
    - Důležitý pojem u 1. úlohy.
    - Graf, kde na jedné ose je poloha a na druhé rychlost.
    - Umožňuje vidět stabilitu systému.
